```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
```

```python
data = {
    'Age' : [25,30,35,40,45,50],
    'Income' : [30000, 40000, 50000, 60000,70000,80000],
    'Spending' : [70,60,50,40,30,20],
    'Savings' : [1000,5000,8000,10000,15000,20000]
}
```

```python
df = pd.DataFrame(data)
```

```python
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)
```

```python
pca = PCA(n_components=2)
pca_df = pca.fit_transform(scaled_data)
```

```python
pca_data = pd.DataFrame(data=pca_df, columns=['PC1', 'PC2'])
explained_variance = pca.explained_variance_ratio_
print('Variance captured by each PCA component')
print(np.round(explained_variance * 100, 2))

plt.figure(figsize=(10,8))
plt.scatter(pca_data['PC1'], pca_data['PC2'])
plt.xlabel('PC1 Main Pattern')
plt.ylabel('PC2 Minor Pattern')
plt.title('PCA Projection (2D-View)')
plt.grid(True)
plt.show()
```
